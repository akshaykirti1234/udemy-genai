# Tuple Operations

```python
numbers = (1,2,3,4,5,6)
mixed_tuple = ("akshay",3,4,5,True,4.5)

print(type(numbers))
print(type(mixed_tuple))
```

```python
concadinate_tuple =  (numbers + mixed_tuple)

print(concadinate_tuple)
```

```python
print(mixed_tuple*3)
```

# Packing and unpacking tuple

```python
# Packing a tuple
packed_tuple = 1,"hello",3.4
print(packed_tuple)
```

```python
# Unpacking a tuple

a,b,c = packed_tuple

print(a)
print(b)
print(c)
```

```python
# Unpacking with *
numbers = (1,2,3,4,5,6,7,8)
first,*middle,last = numbers

print(first)
print(middle)
print(last)
```

```python
# Nested List

lst = [[1,2,3,4],[5,6,7,8],[1,'hello',3.4,c]]

print(lst)

lst[0][3]
```

```python
# Nested Tuples

nested_tuple = ((1,2,3),(4,5,6),(1,"hello",3.4),(True,False))
```

```python
print(nested_tuple)
```

```python
print(nested_tuple[1])
```

```python
print(nested_tuple[2][1])
```

```python
# Iterating over the nested_tuple

for sub_tuple in nested_tuple:
    for item in sub_tuple:
        print(item,end=" ")
    print()
```
