### Dictionaries are unordered collections of items. They store data in key-value pairs. Keys must be unique and immutable(e.g strings, numbers, or inputs) while values can be of many types.

```python
# Creating Dictionaries
empty_dict={}

print(type(empty_dict))
```

```python
empty_dict = dict()
empty_dict
```

```python
student = {"name":"Akshay","age":26,"grade":24}
print(student)
print(type(student))
```

```python
# Single key is always used
student = {"name":"Akshay","age":26,"name":24}
print(student)  
```

```python
# accessing dictionary elements
student = {"name":"Akshay","age":26,"grade":"A"}

print(student['grade'])
print(student['age'])
print(student['name'])
```

```python
# Accessing using get() method

print(student.get('grade'))
print(student.get('name'))
print(student.get('age'))
print(student.get('last_name'))
print(student.get('last_name','Not Availabe'))
```

```python
# Modifying Dictionary Elements
# Dictionary are mutable, so you can add, update or delete elements

print(student)
```

```python
student['age'] = 20
```

```python
print(student)
```

```python
# Add a new key
student['address'] = 'India'
```

```python
print(student)
```

```python
# Delete a key
del student['grade']
```

```python
print(student)
```

```python
# Dictionary methods

keys = student.keys()  # Get all the keys
print(keys)

values = student.values() # Get all the values
print(values)

items = student.items() # Get all key value pairs
print(items)
```

# Shallow copy

```python
# Shallow copy

student_copy = student

print(student)
print(student_copy)
```

```python
student['name']= 'Akshay2'

print(student)
print(student_copy)
```

```python
student_copy1 = student.copy() # Shallow copy
print(student_copy1)
print(student)
```

```python
student['name'] = "Akshay3"
print(student)
print(student_copy1)
```

# Iterating over dictionaries

```python
for keys in student.keys():
    print(keys)
```

```python
# Iterate over values
for values in student.values():
    print(values)
```

```python
# Iterate over key-value pairs
for key,val in student.items():
    print(f"{key}:{values}")
```

```python
# Nested Dictionaries
student = {
    "sudent1": {"name":'Akshay',"age":26},
    "sudent2": {"name":'Subrat',"age":25},
    "sudent3": {"name":'Gopinath',"age":22},
}
```

```python
for key,val in student.items():
    print(f"{key}:{val}")
```

```python
# Access nested dictionaries element
print(student['sudent1']['name'])
print(student['sudent1']['age'])
```

```python
# Iterating over nested dictionaries
for student_id,student_info in student.items():
    print(f'{student_id}:{student_info}')
    for key,val in student_info.items():
        print(f'{key}:{val}')
    print()
```

# Dictionary comprehension

```python
squares = {x:x**2 for x in range(5)}

print(squares)
```

# Conditional dictionary comprehension

```python
evenSquares = {x:x**2 for x in range(5) if x%2 == 0}

print(evenSquares)
```

```python
# Use a dictionary to count the frequency of elements in a list.

number_list = [1,2,3,2,1,4,3,2,3,3,4]
number_frequency = {}

for num in number_list:
    if num in number_frequency:
        number_frequency[num]+=1
    else:
        number_frequency[num] = 1
        
print(number_frequency)
```

```python
# Merge two dictionaries into one

dict1 = {"a":1,"b":2}
dict2 = {"b":3,"c":4}

merged_dict = {**dict1,**dict2}

print(merged_dict)
```
