#### Example 1 : Tempratue Conversion

```python
def convert_temprature(temp,unit):
    """This function converts temprature between celseus and fahrenheit"""
    if unit == "C":
        return temp * 9/5 +32
    elif unit == "F":
        return (temp-32)*5/9
    else:
        return None
```

```python
print(convert_temprature(25,'C'))
print(convert_temprature(77,'F'))
```

#### Example 2 : Password Strength Checker

```python
def is_strong_password(password):
    """This function checks wheather the password is strong or not"""
    if len(password) < 8:
        return False
    if not any(char.isdigit() for char in password):
        return False
    if not any(char.islower() for char in password):
        return False
    if not any(char.isupper() for char in password):
        return False
    if not any(char in '!@#$%^&()_+' for char in password):
        return False
    return True
    
```

```python
print(is_strong_password("WeakPass"))
print(is_strong_password("Pass@123"))
```

#### Example 3 : Calculate the total cost of items in a shopping cart

```python
def calculate_total_cost(cart):
    total_cost = 0
    for item in cart:
        total_cost += item['price'] * item['qty']
    return total_cost


# Example Cart
cart = [
    {'name':'Apple','price':0.5,'qty':4},
    {'name':'Banana','price':0.3,'qty':6},
    {'name':'Orange','price':0.7,'qty':3}
]
```

```python
calculate_total_cost(cart)
```

#### Example 4 : check if a string is pallindrome

```python
def is_pallimrome(str):
    str = str.lower().replace(" ","")
    return str == str[::-1]
```

```python
print(is_pallimrome("A man a plan a canal panama"))
(is_pallimrome("Hello"))
```

#### Example 5 : Calculate the factorials of a number using recursion

```python
def factorial(n):
    if n == 0:
        return 1
    else:
        return n * factorial(n-1)
```

```python
factorial(5)
```

#### Exsample 6: A Fucntino to read a file and count the frequency of each word

```python
def count_word_frequency(file_path):
    word_count = {}
    with open(file_path,'r') as file:
        for line in file:
            words = line.split()
            for word in words:
                word = word.strip('!@#$%^&*()-_=+,.?')
                word_count[word] = word_count.get(word,0)+1
    return word_count
```

```python
filePath = 'sample.txt'
count_word_frequency(filePath)
```
