#### Example 1 : Tempratue Conversion

```python
def convert_temprature(temp,unit):
    """This function converts temprature between celseus and fahrenheit"""
    if unit == "C":
        return temp * 9/5 +32
    elif unit == "F":
        return (temp-32)*5/9
    else:
        return None
```

```python
print(convert_temprature(25,'C'))
print(convert_temprature(77,'F'))
```
