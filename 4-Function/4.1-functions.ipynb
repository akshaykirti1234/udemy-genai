```python
# Why fucntions?
num = 24
if num%2 == 0:
    print("Number is Even")
else:
    print("Number is Odd")
    
```

```python
def even_or_odd(num):
    """This function is finds even or odd"""
    if num%2 == 0:
        print("Number is Even")
    else:
        print("Number is Odd")
```

```python
# Call the fucntion
even_or_odd(5)
```

```python
# Variable lenght arguments
# Positional and Keyword arguments

# Positional arguments
def print_numbers(*args):
    for numbers in args:
       print(numbers) 
```

```python
print_numbers(1,2,3,4,5,'akshay',True,False,9.0)
```

```python
# Keywords argumemts

def print_details(**kwargs):
    for key,val in kwargs.items():
        print(f"{key}:{val}")
```

```python
print_details(name="Akshay",age="26",country="India")
```
