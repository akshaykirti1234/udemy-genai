```python
# While Loop

count = 0

while count < 10:
    print(count)
    count += 1
```

```python
# Break
for i in range(10):
    if i == 5:
        break
    print(i)
```

```python
# Continue
for i in range(10):
    if i%2 == 0:
        continue
    print(i)
```

```python
# Pass
# ----------
# pass is a null statement in Python. It does nothing and is used as a placeholder where a statement is required syntactically.

for i in range(5):
    if i == 3:
        pass
    print(i)
```

```python
for i in range(1,3):
    for j in range(1,5):
        print(f"i: {i} and j: {j}")
```

```python
n = 10
sum = 0
count = 1

while count <= n:
    sum = sum+count
    count += 1
print(f"Sum: {sum}")
```

```python
# Using for loop

n = 10
sum = 0

for i in range(n+1):
    sum += i
print(sum)
```

```python
# Prime Number

for num in range (1,101):
    if num > 1:
        for i in range(2,num):
            if(num % i == 0):
                break
            else:
                print(num)
                break
```
